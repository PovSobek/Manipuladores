# Fórmulas de Par Gravitacional Actualizadas

La presencia de la masa del efector y de la carga ($m_{efector}, m_{carga}$), ambas situadas al final del brazo, modifica nuestras ecuaciones de energía potencial. Por tanto, las fórmulas para los pares gravitacionales se actualizan de la siguiente manera:

$$ \tau_1 = \left(\frac{m_1}{2} + m_2 + m_{efector} + m_{carga}\right) g L_1 \cos(\theta_1) + \left(\frac{m_2}{2} + m_{efector} + m_{carga}\right) g L_2 \cos(\theta_1 + \theta_2) $$

$$ \tau_2 = \left(\frac{m_2}{2} + m_{efector} + m_{carga}\right) g L_2 \cos(\theta_1 + \theta_2) $$

### **Ejercicio Práctico:**

Adapta la función de la celda siguiente para que utilice estas nuevas fórmulas y los parámetros interactivos que acabas de introducir.

```python
def calcular_par_gravedad_completo(theta1_grados, theta2_grados, L1, L2, m1, m2, m_efector, m_carga, g):
    """
    Calcula el par motor incluyendo las masas del efector final y la carga.
    """
    th1 = np.deg2rad(theta1_grados)
    th2 = np.deg2rad(theta2_grados)

    # Agrupamos las masas para simplificar
    m_final = m_efector + m_carga


    # ▼▼▼ TU CÓDIGO AQUÍ ▼▼▼

    # Implementa las nuevas fórmulas para tau1 y tau2.
    # Puedes crear términos intermedios para que sea más legible.


    tau2 = (m2 / 2 + m_final) * g * L2 * np.cos(th1 + th2) # Reemplaza el 0 con la fórmula correcta
    tau1 = (m1 / 2 + m2 + m_final) * g * L1 * np.cos(th1) + tau2# Reemplaza el 0 con la fórmula correcta

    # ▲▲▲ FIN DE TU CÓDIGO ▲▲▲


    return (tau1, tau2)

# --- Probamos tu función ---
# Mantenemos el brazo extendido horizontalmente
th1_test, th2_test = 0, 0
par_calculado = calcular_par_gravedad_completo(th1_test, th2_test, L1, L2, m1, m2, m_efector, m_carga, g)

# El resultado ahora dependerá de los valores que introdujiste
print(f"Para mantener la posición horizontal ({th1_test}°, {th2_test}°):")
print(f"  - Par en Articulación 1 (τ1): {par_calculado[0]:.2f} Nm")
print(f"  - Par en Articulación 2 (τ2): {par_calculado[1]:.2f} Nm")
```

## Análisis Experimental

Cópia e introduce el código insertado anteriormente. La visualización será la misma que en el cuaderno anterior, pero ahora es mucho más interesante: puedes volver a la celda 2, cambiar los parámetros (por ejemplo, duplicar el peso de la carga o hacer un brazo más largo) y volver a ejecutar esta celda de solución para ver **instantáneamente** cómo afecta eso a los motores.

```python
def calcular_par_gravedad_completo_solucion(theta1_grados, theta2_grados, L1, L2, m1, m2, m_efector, m_carga, g):
    th1 = np.deg2rad(theta1_grados)
    th2 = np.deg2rad(theta2_grados)
    m_final = m_efector + m_carga

   # ▼▼▼ TU CÓDIGO AQUÍ ▼▼▼
    # Implementa las nuevas fórmulas para tau1 y tau2.
    # Puedes crear términos intermedios para que sea más legible.

    tau2 = (m2 / 2 + m_final) * g * L2 * np.cos(th1 + th2) # Reemplaza el 0 con la fórmula correcta
    tau1 = (m1 / 2 + m2 + m_final) * g * L1 * np.cos(th1) + tau2# Reemplaza el 0 con la fórmula correcta

    # ▲▲▲ FIN DE TU CÓDIGO ▲▲▲


    return (tau1, tau2)

# --- VISUALIZACIÓN ---

print("--- Introduce el ángulo fijo para el primer eslabón ---")
theta1_fijo = float(input("Introduce el ángulo Theta1 (grados) para el análisis (ej: 45): "))
# Creamos un rango de valores para theta2
theta2_rango = np.linspace(-180, 180, 500)

lista_tau1, lista_tau2 = [], []

# Calculamos los pares para cada valor de theta2 con los parámetros introducidos por el usuario
for th2_grados in theta2_rango:
    params = (theta1_fijo, th2_grados, L1, L2, m1, m2, m_efector, m_carga, g)
    tau1_res, tau2_res = calcular_par_gravedad_completo_solucion(*params)
    lista_tau1.append(tau1_res)
    lista_tau2.append(tau2_res)

# Gráfico
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(10, 8), sharex=True)
# Título dinámico que refleja los parámetros clave
fig.suptitle(f'Análisis de Pares con m1={m1}kg, m2={m2}kg, Carga={m_carga}kg', fontsize=16)

ax1.plot(theta2_rango, lista_tau1, 'r-')
ax1.set_title(f'Theta1 fijo en {theta1_fijo}°')
ax1.set_ylabel('Par τ1 (Nm)')
ax1.grid(True)

ax2.plot(theta2_rango, lista_tau2, 'b-')
ax2.set_xlabel('Ángulo Theta2 (grados)')
ax2.set_ylabel('Par τ2 (Nm)')
ax2.grid(True)

plt.tight_layout(rect=[0, 0.03, 1, 0.95])
plt.show()
```

Este código analiza y visualiza las fuerzas que un brazo robótico de dos eslabones necesita para sostenerse a sí mismo y a una carga en contra de la gravedad.

En esencia, responde a la pregunta: **"¿Cuánta fuerza (torque) deben hacer los motores para que el brazo no se caiga en una posición determinada?"**

El código se divide en dos partes principales:

---
### 1. La Función de Cálculo (`calcular_par_gravedad_completo_solucion`)

Esta función es el "cerebro" del análisis. Realiza los siguientes pasos:
* **Recibe los parámetros**: Acepta la posición del robot (los ángulos de sus dos articulaciones en grados) y sus características físicas (longitud y masa de los eslabones, masa de la pinza y de la carga que sostiene).
* **Convierte a radianes**: Las funciones trigonométricas en Python (seno, coseno) trabajan con radianes, por lo que convierte los ángulos de entrada.
* **Calcula el torque para el Motor 1 (`tau1`)**: Calcula la fuerza de rotación que el primer motor (el que está en la base) necesita para soportar el peso de **todo** lo que viene después de él:
    * El peso del primer eslabón.
    * El peso del segundo eslabón.
    * El peso de la pinza y la carga.
    Usa el coseno de los ángulos para determinar cómo la gravedad "tira" del brazo dependiendo de su postura.
* **Calcula el torque para el Motor 2 (`tau2`)**: De forma similar, calcula la fuerza que el segundo motor (el "codo") necesita para soportar el peso de lo que viene después de él:
    * El peso del segundo eslabón.
    * El peso de la pinza y la carga.
* **Devuelve los resultados**: Entrega los dos valores de torque calculados.

---
### 2. La Sección de Visualización

Esta parte del código utiliza la función anterior para crear un análisis visual muy útil.
* **Fija una posición**: Mantiene el primer eslabón del brazo en una posición fija de 45 grados (`theta1_fijo = 45.0`).
* **Crea un movimiento**: Genera una secuencia de 500 ángulos para el segundo eslabón, cubriendo un giro completo de -180 a 180 grados (`theta2_rango`).
* **Realiza un barrido**: Entra en un bucle (`for`) y, para cada una de las 500 posiciones del segundo eslabón, llama a la función de cálculo para saber qué torques se necesitan. Guarda estos 500 resultados en dos listas (`lista_tau1` y `lista_tau2`).
* **Genera los gráficos**: Al final del bucle, crea dos gráficos:
    * **Gráfico Superior**: Muestra cómo cambia el esfuerzo del **Motor 1** a medida que el segundo eslabón se mueve. El eje X es el ángulo del segundo eslabón y el eje Y es el torque necesario para el primer motor.
    * **Gráfico Inferior**: Hace lo mismo pero para el **Motor 2**.

El resultado final son dos gráficos que permiten a un ingeniero entender **en qué posiciones los motores tienen que trabajar más** y, por lo tanto, ayuda a decidir qué tan potentes deben ser los motores para el robot. 🦾
