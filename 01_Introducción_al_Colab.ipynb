# <font color='grey'>**7. Generar secciones/encabezados/índices**


En los documentos se pueden ir estructurando por niveles de secciones/encabezados.
Se utiliza el símbolo `#` para indicar cada nivel del encabezado.

Para que quede establecido y poder reproducir un índice, cada encabezado/sección ha de definirse en una celda distinta al cuerpo del contenido. Si accedes al símbolo del listado en la barra lateral, accederás al ínidce creado con este sistema.

```markdown
# Sección/encabezado 1
## Subsección/encabezado de la Sección/encabezado  1
### Subsección/encabezado de la subsección/encabezado  de la Sección 1
# Sección/encabezado  2
# Sección 3
```


# <font color='grey'>**8. Listas ordenadas y no ordenadas**

En las celdas de texto de Colab, las listas se crean usando la sintaxis de Markdown para estructurar la información de forma clara y jerárquica.
<br><br>

Hay dos tipos principales:

**Listas con Símbolos (No Ordenadas):** Se usan para elementos donde el orden no

importa (como una lista de ingredientes o features).

**Listas Numéricas (Ordenadas):** Se usan para elementos que deben seguir una secuencia específica (como pasos, instrucciones o clasificaciones).

Tipo de Lista	Símbolos Permitidos.

  **No Ordenada:**	Asterisco (*), guion (-) o signo de más (+).
El símbolo debe ir seguido de un espacio antes del texto.

  **Ordenada:**	Un número seguido de un punto (.)

Puedes usar 1., 2., 3., etc.

El número que uses al inicio no es tan importante; Markdown lo ordenará automáticamente, pero se recomienda empezar con 1. para mayor claridad.

# <font color='grey'> **9. Insertar Imágenes en Celdas de Texto**

En Google Colab, la forma más común de insertar imágenes en una celda de texto es usando la sintaxis de Markdown o simplemente arrastrando y soltando el archivo.
<br><br>

**Mecanismo: Markdown**

La sintaxis de Markdown para imágenes tiene dos partes: un texto alternativo y la ruta (o URL) de la imagen.


* ! [Texto descriptivo de la imagen] (URL o Ruta de la imagen)


**!**	Indica que se va a insertar una imagen.

* **[Texto descriptivo]**	Es el texto alternativo. Es lo que se muestra si la imagen no carga y es importante para la accesibilidad.

* **(URL o Ruta)**	Es la dirección web de la imagen (si está en internet) o la ruta local dentro de Colab.

Esta es la sintaxis para mostrar una imagen desde la web:

"![Logo de Python] (https://www.python.org/static/community_logos/python-logo-master-v3-TM-stacked.png)"

**Mecanismo: Arrastrar y Soltar**

La manera más fácil es:

* Abre una celda de texto en Colab.

* Arrastra un archivo de imagen (JPEG, PNG, etc.) desde tu computadora y suéltalo en la celda.

* Colab subirá la imagen a un directorio temporal del cuaderno y generará automáticamente la sintaxis Markdown con la ruta correcta.

# <font color='grey'> **10. Uso de LaTeX**



Vamos a insertar ecuaciones matemáticas de manera elegante y profesional en tus documentos de Colab, usando la sintaxis LaTeX.

El formato LaTeX es estándar en el mundo científico y académico, y Colab lo renderiza automáticamente dentro de las celdas de texto.
<br><br>
**1. Látex Básico**

Ecuaciones Inline y en Bloque (Centradas)

La sintaxis básica de LaTeX se define mediante delimitadores:

Ecuaciones Inline: Se utilizan para incrustar fórmulas dentro del flujo normal del texto. Se delimitan con un solo signo de dólar a cada lado: "$".

Ecuaciones en Bloque (Display/Centradas): Se utilizan para mostrar ecuaciones importantes en su propia línea y centradas. Se delimitan con doble signo de dólar a cada lado: "$$".

Ejemplo

La famosa ecuación cuadrática $ax^2 + bx + c = 0$

se escribe inline usando \\$ax^2 + bx + c = 0\\$.
<br><br>
La solución a esta ecuación, mostrada en bloque, sería:
<br><br>
$$x = \frac{-b \pm \sqrt{b^2 - 4ac}}{2a}$$
<br><br>
Esta ecuación en bloque se creó con \\$\\$ x = \frac{-b \pm \sqrt{b^2 - 4ac}}{2a} \$$

**2. Exponentes y Subíndices**



**Exponentes:** Se indican con el símbolo de intercalación `^`.

**Subíndices:** Se indican con el guion bajo `_`.

**Agrupación:** Si el exponente o subíndice tiene más de un carácter, se deben agrupar con llaves `{}`.

Ejemplos de sintaxis:

x^2: $x^2$


x_{ij}: $x_{ij}$

e^{i \pi}: $e^{i \pi}$
  (Usamos {} para agrupar iπ)
<br><br>
**Ejemplo**

La fórmula de la Energía Cinética:

$E\_k = \frac{1}{2} m v^2$


Se escribe como \\$\\$E\_k = \frac{1}{2} m v^2\\$\\$

**3. Funciones y Símbolos Comunes**


LaTeX proporciona comandos con barra invertida (\) para símbolos y funciones matemáticas comunes:

Función | Código | Denominación
--- | --- | ---
$\frac{a}{b}$ | frac{a}{b} | Fracción
$\sqrt{x}$ | \sqrt{x} | Raíz cuadrada
$\sum_{i=0}^n$ | \sum_{i=0}^n | Sumatoria con límites
$\int_a^b$ | \int_a^b	| Integral con límites
$\sin(x)$	| \sin(x)	| Función trigonométrica
$\pi$ $\alpha$$ |	\pi,\alpha | Letras griegas

<br><br>
Ejemplo

Una identidad integral definida:
<br><br>
$$\int_0^1\frac{dx}{1+x^2} = \frac{\pi}{4}$$
<br><br>
Se escribe como \\$\\$\int_0^1 \frac{dx}{1+x^2} = \frac{\pi}{4}\\$\\$
<br><br>

**4. Matrices y Delimitadores (Paréntesis Ajustables)**

Las matrices se crean usando el entorno array o entornos especializados como **pmatrix** (matriz con paréntesis) dentro de  `$$...$$`

- **begin{pmatrix}** / **end{pmatrix}**: crea una matriz entre paréntesis grandes.


- Los elementos se separan por `&`.


- Las filas se separan por doble barra invertida `\\`.

Además, para que los paréntesis se ajusten al tamaño de fracciones o matrices, se usan los comandos `\left` y `\right`.
<br><br>
Ejemplo

La Matriz de Rotación 2D:
$$
R(\theta) =
\begin{pmatrix}
\cos \theta & -\sin \theta \\
\sin \theta & \cos \theta
\end{pmatrix}
$$

Se escribe como
<br><br>
```latex
$$R(\theta) =\begin{pmatrix}\cos \theta & -\sin \theta \\\sin \theta & \cos \theta\end{pmatrix}$$
```


**5. Centrar Ecuaciones con Etiquetas Multi-Línea**

Aunque \\$\\$...\\$\\$ centra automáticamente una ecuación, para alinear varias ecuaciones o para numerarlas, se usan entornos más avanzados:

**Alineación Multi-Línea (align):**
El entorno $\begin{align}$ y $\end{align}$ permite alinear varias ecuaciones usando el símbolo **&** como punto de alineación (generalmente antes de un signo de igualdad **=**

Ejemplo
Alineación de un sistema de ecuaciones:

\begin{align} 2x + 3y &= 5 \\ 4x - y &= 3 \end{align}

El código usado es
```latex
\begin{align} 2x + 3y &= 5 \\ 4x - y &= 3 \end{align}
```
