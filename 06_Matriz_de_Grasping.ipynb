# Matriz de Agarre Interactiva 💪

Para el estudio de la matriz de Grasping, el programa te pedirá los parámetros del objeto y del agarre, y luego calculará la Matriz de Agarre y las fuerzas necesarias para la tarea.

**Conceptos Clave:**
* **Fuerzas de Contacto ($f_c$)**: Las fuerzas que ejercen los dedos.
* **Wrench (Vorticidad, $w$)**: El efecto neto de fuerza y par motor sobre el objeto.
* **Matriz de Agarre (G)**: La matriz que conecta ambos conceptos a través de la ecuación $w = G \cdot f_c$.

Vamos a realizar un agarre sin fricción

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle

# --- Configuración Interactiva del Escenario ---
try:
    print("--- Introduce los PARÁMETROS DEL OBJETO ---")
    ancho_obj = float(input("Ancho del objeto en metros (ej: 0.1): "))
    alto_obj = float(input("Alto del objeto en metros (ej: 0.2): "))
    masa_obj = float(input("Masa del objeto en kg (ej: 0.5): "))

    print("\n--- Introduce los PARÁMETROS DEL AGARRE (2 Dedos) ---")
    print("El origen (0,0) está en el centro del objeto.")

    # Dedo 1
    p1_x = float(input("Posición X del contacto 1 (ej: -0.05): "))
    p1_y = float(input("Posición Y del contacto 1 (ej: 0): "))
    angulo1_grados = float(input("Ángulo de la fuerza 1 (grados, 0 es a la derecha, ej: 0): "))

    # Dedo 2
    p2_x = float(input("\nPosición X del contacto 2 (ej: 0.05): "))
    p2_y = float(input("Posición Y del contacto 2 (ej: 0): "))
    angulo2_grados = float(input("Ángulo de la fuerza 2 (grados, ej: 180): "))

    # Ensamblamos los datos de posición
    p1 = np.array([p1_x, p1_y])
    p2 = np.array([p2_x, p2_y])
    puntos_contacto = [p1, p2]

    # Convertimos los ángulos a vectores normales unitarios
    angulo1_rad = np.deg2rad(angulo1_grados)
    angulo2_rad = np.deg2rad(angulo2_grados)
    n1 = np.array([np.cos(angulo1_rad), np.sin(angulo1_rad)])
    n2 = np.array([np.cos(angulo2_rad), np.sin(angulo2_rad)])
    normales_contacto = [n1, n2]

    print("\n¡Escenario de agarre configurado con éxito!")

except ValueError:
    print("\nError: Por favor, introduce solo valores numéricos. Vuelve a ejecutar la celda.")
```

<br><br>
$$ g_i = \begin{bmatrix} n_i \\ p_i \times n_i \end{bmatrix} $$
<br><br>
Como ahora los puntos `p` y las normales `n` los defines tú, la matriz `G` cambiará en cada ejecución.

## Aplicación: ¿Qué fuerzas necesito para levantar el objeto?

Usaremos la pseudoinversa de la matriz `G` que acabas de calcular para resolver el problema inverso: dado un `wrench_deseado` (levantar el objeto contra la gravedad), calcularemos las `fuerzas_contacto` necesarias.
<br><br>
$$ f_c = G^{+} w_{deseado} $$
<br><br>
La celda siguiente realizará este cálculo y visualizará el resultado basado en los parámetros que definiste.


**Convenio de Ángulos:**
* **0°** apunta hacia la derecha (eje X positivo).
* **90°** apunta hacia arriba (eje Y positivo).
* **180°** apunta hacia la izquierda.
* **270°** o **-90°** apunta hacia abajo.

El código convertirá automáticamente estos ángulos en los vectores normales necesarios para construir la Matriz de Agarre.

La función para construir la matriz `G` está diseñada para trabajar con vectores normales, que calcularemos a partir de los ángulos.

En la celda siguiente, no solo calcularemos las fuerzas para **levantar el objeto**, sino que también te pediremos un **torque deseado** para que puedas probar una nueva visualización. Para vencer sólo el peso (la fuerza de la gravedad) introduciremos un **torque** de valor 0.


```python
def construir_matriz_grasp(puntos_contacto, normales_contacto):
    """
    Construye la Matriz de Agarre (G). Esta función no necesita cambios.
    """
    columnas_g = []
    for p, n in zip(puntos_contacto, normales_contacto):


        # --- ▼▼▼ TU CÓDIGO AQUÍ ▼▼▼ ---

        # 1. Calcula el componente de par motor (torque) para este contacto.
        torque = 0 #sustituye el 0 por la formulación correcta

        # 2. Crea la columna para este dedo.
        columna_actual = 0 #sustituye el 0 por la formulación correcta

        # --- ▲▲▲ FIN DE TU CÓDIGO ▲▲▲ ---


        columnas_g.append(columna_actual)
    return np.hstack(columnas_g)

# --- APLICACIÓN CON DATOS INTERACTIVOS ---

try:
    # 1. Construir la matriz G
    G = construir_matriz_grasp(puntos_contacto, normales_contacto)
    print("Matriz de Agarre (G):")
    print(G)

    # 2. Definir el wrench deseado
    fuerza_gravedad = masa_obj * 9.81
    torque_deseado_input = float(input("\nIntroduce un torque deseado (Nm, ej: 0 para solo levantar): "))
    wrench_deseado = np.array([0, fuerza_gravedad, torque_deseado_input])

    # 3. Calcular la pseudoinversa y las fuerzas de contacto
    G_pseudoinversa = np.linalg.pinv(G)
    fuerzas_contacto = G_pseudoinversa @ wrench_deseado
    print(f"\nWrench Deseado: {wrench_deseado}")
    print(f"Fuerzas de Contacto calculadas (f1, f2): {fuerzas_contacto} N")

    # --- VISUALIZACIÓN ---
    fig, ax = plt.subplots(figsize=(6,8)); ax.set_aspect('equal')
    obj = Rectangle((-ancho_obj/2, -alto_obj/2), ancho_obj, alto_obj, color='lightblue', label='Objeto')
    ax.add_patch(obj)

    # Dibujar fuerzas de contacto solo si son significativas
    fuerzas_son_cero = True
    for i, p in enumerate(puntos_contacto):
        fuerza_magnitud = fuerzas_contacto[i]

        if abs(fuerza_magnitud) > 1e-6:
            fuerzas_son_cero = False
            normal = normales_contacto[i]

            # Define una longitud visual para la flecha (para que no sea demasiado grande/pequeña)
            largo_visual_flecha = 0.1

            # El vector de la flecha tiene la dirección de la normal y una longitud visual fija
            dx = normal[0] * largo_visual_flecha
            dy = normal[1] * largo_visual_flecha

            # El inicio de la flecha (la cola) es el punto de contacto 'p' MENOS el vector de la flecha
            inicio_x = p[0] - dx
            inicio_y = p[1] - dy

            # Dibuja la flecha desde la cola (inicio) con el vector (dx, dy)
            # La punta terminará exactamente en (inicio_x + dx, inicio_y + dy) = p
            ax.arrow(inicio_x, inicio_y, dx, dy,
                     head_width=0.02, head_length=0.03, fc='red', ec='red', length_includes_head=True)

            # Posicionar el texto cerca de la cola de la flecha
            ax.text(inicio_x - 0.02, inicio_y - 0.02,
                    f'f{i+1}={fuerza_magnitud:.2f}N', color='red', fontsize=9)

    if fuerzas_son_cero:
        ax.text(0, 0, "ADVERTENCIA:\nLa tarea es físicamente imposible\ncon este agarre.",
                 ha='center', va='center', color='darkred', fontsize=12,
                 bbox=dict(facecolor='white', alpha=0.8))

    # Dibujar wrench (Fuerza)
    ax.arrow(0, 0, 0, 0.1, head_width=0.02, head_length=0.03, fc='green', ec='green', length_includes_head=True, label='Fuerza Vertical Deseada')

    # Visualización del Torque
    torque_deseado = wrench_deseado[2]
    if not np.isclose(torque_deseado, 0):
        ax.plot(0, 0, 'o', ms=16, mfc='none', mec='purple', label=f'Torque: {torque_deseado:.2f} Nm')
        if torque_deseado > 0: ax.plot(0, 0, '.', ms=10, color='purple')
        else: ax.plot(0, 0, 'x', ms=12, color='purple', mew=2)

    margen = max(ancho_obj, alto_obj) * 0.7
    ax.set_xlim(-ancho_obj/2 - margen, ancho_obj/2 + margen)
    ax.set_ylim(-alto_obj/2 - margen, alto_obj/2 + margen)
    ax.grid(True); ax.legend(); ax.set_title("Análisis de Fuerzas de Agarre")
    plt.show()

except NameError:
    print("Por favor, ejecuta primero la celda de configuración interactiva (Celda 2).")
except ValueError:
    print("\nError: Por favor, introduce un valor numérico para el torque.")
```
